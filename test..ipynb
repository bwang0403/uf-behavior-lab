```python
import faster_whisper, sounddevice, soundfile, time

# 录3秒
audio = sounddevice.rec(int(3 * 16000), samplerate=16000, channels=1)
sounddevice.wait()
soundfile.write("test.wav", audio, 16000)

# 测识别速度
model = faster_whisper.WhisperModel("base", device="cpu")
t0 = time.time()
segments, _ = model.transcribe("test.wav", language="en")
print(f"识别耗时: {time.time()-t0:.2f}秒")
print([s.text for s in segments])
```
